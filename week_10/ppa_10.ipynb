```c
void createFile(int arr[],int n){
    FILE *file;
    file=fopen("output.txt","w");
    if (file!=NULL){
        for (int i=0;i<n;i++){
            fprintf(file,"%d\n",arr[i]);
        }
    }
    
    fclose(file);
}
```
